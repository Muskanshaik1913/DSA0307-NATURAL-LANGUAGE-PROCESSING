# Experiment 1 – Context-Free Grammar (CFG) Validation and Parse Tree Construction

## Experiment Description
Validate whether an input sentence satisfies the given CFG production rules and construct its parse tree.

### Grammar
- S → NP VP
- NP → Det N
- VP → V NP
- Det → the | a
- N → student | teacher | book
- V → reads | likes

## Aim
To validate English sentences using CFG production rules and generate parse tree representations.

## Algorithm
1. Define the CFG grammar.
2. Create a Recursive Descent Parser.
3. Accept a sentence from the user.
4. Parse the sentence.
5. If valid, display the parse tree.
6. Otherwise display **Invalid Sentence**.

## Function Prototype
```python
def parse_cfg(sentence):
```


````python
from nltk import CFG
from nltk.parse import RecursiveDescentParser
grammar = CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the' | 'a'
N -> 'student' | 'teacher' | 'book'
V -> 'reads' | 'likes'
""")
parser = RecursiveDescentParser(grammar)
def parse_cfg(sentence):
    words=sentence.split()
    try:
        trees=list(parser.parse(words))
        if trees:
            print("Valid Parse Tree\n")
            for tree in trees:
                print(tree)
                try:
                    tree.pretty_print()
                except:
                    pass
        else:
            print("Invalid Sentence")
    except Exception:
        print("Invalid Sentence")
user=input("Enter a sentence: ")
parse_cfg(user)
print("\nOfficial Test Cases")
tests=[
"the student reads a book",
"a teacher likes the book",
"student reads book",
"the book likes a teacher",
"reads the student book"
]
for t in tests:
    print("\nInput:",t)
    parse_cfg(t)
````

## Expected Output
- the student reads a book → Valid Parse Tree
- a teacher likes the book → Valid Parse Tree
- student reads book → Invalid Sentence
- the book likes a teacher → Valid Parse Tree
- reads the student book → Invalid Sentence

## Analysis
The parser accepts only sentences generated by the CFG. Invalid sentence structures are rejected.

## Result
The program successfully validates sentences using CFG production rules and constructs parse tree representations for valid sentences.


# Experiment 2 – Agreement and Feature Structure Checking

## Experiment Description
Verify subject–verb agreement using feature structures containing Number and Person attributes.

## Aim
To verify grammatical agreement between subjects and verbs.

## Algorithm
1. Define feature structures.
2. Accept subject and verb.
3. Compare their features.
4. Return True if agreement holds; otherwise False.

## Function Prototype
```python
def check_agreement(subject, verb):
```


````python
subjects={
'he':('singular','third'),
'she':('singular','third'),
'it':('singular','third'),
'they':('plural','third')
}

verbs={
'runs':('singular','third'),
'writes':('singular','third'),
'run':('plural','third'),
'write':('plural','third')
}

def check_agreement(subject,verb):
    if subject not in subjects or verb not in verbs:
        return False
    return subjects[subject]==verbs[verb]

s=input("Enter subject: ")
v=input("Enter verb: ")
print("Agreement:",check_agreement(s,v))

print("\nOfficial Test Cases")
for s,v in [('he','runs'),('he','run'),('they','run'),('they','writes'),('she','writes')]:
    print(s,v,'->',check_agreement(s,v))
````

## Expected Output
- he runs → True
- he run → False
- they run → True
- they writes → False
- she writes → True

## Analysis
The feature structures compare number and person attributes to determine agreement.

## Result
The program correctly verifies subject–verb agreement.


# Experiment 3 – Probabilistic Context-Free Grammar (PCFG)

## Experiment Description
Parse a two-word sentence and compute its probability using the given PCFG.

## Aim
To compute sentence probabilities using probabilistic grammar rules.

## Algorithm
1. Store NP and VP probabilities.
2. Accept a two-word sentence.
3. Verify that both words exist.
4. Compute:

**P(Sentence)=P(S→NP VP)×P(NP)×P(VP)**

5. Return 0 for invalid sentences.

## Function Prototype
```python
def sentence_probability(sentence):
```


````python
np_prob={'John':0.6,'Mary':0.4}
vp_prob={'runs':0.5,'walks':0.5}

def sentence_probability(sentence):
    words=sentence.split()
    if len(words)!=2:
        return 0.0
    if words[0] in np_prob and words[1] in vp_prob:
        return np_prob[words[0]]*vp_prob[words[1]]
    return 0.0

user=input("Enter two-word sentence: ")
print("Probability:",format(sentence_probability(user),'.2f'))

print("\nOfficial Test Cases")
for t in ['John runs','John walks','Mary runs','Mary walks','Peter runs']:
    print(t,'->',format(sentence_probability(t),'.2f'))
````

## Expected Output
- John runs → 0.30
- John walks → 0.30
- Mary runs → 0.20
- Mary walks → 0.20
- Peter runs → 0.00

## Analysis
The probability is calculated using the PCFG production probabilities. Invalid sentences receive a probability of 0.00.

## Result
The program successfully computes sentence probabilities using the given PCFG.
